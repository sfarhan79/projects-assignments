```python
import ipywidgets as widgets
from IPython.display import display, clear_output

def calculate_investment_value(rate, investment_amount, period, is_annuity, annual_amount=None, amounts_per_year=None):
    """
    Calculates the total present value of cash flows and provides an investment recommendation.
    """
    if rate <= 0:
        print("Error: Rate must be greater than zero. Please enter a positive rate.")
        return
    if period <= 0:
        print("Error: Period must be greater than zero. Please enter a positive number of years.")
        return

    total_present_value = 0.0
    r = rate / 100.0 # Convert percentage to decimal

    print(f"\n--- Calculation Details ---")
    print(f"Rate: {rate}% (r = {r:.4f})")
    print(f"Period: {period} years")
    print(f"Initial Investment Amount: {investment_amount:.2f}")

    if is_annuity:
        if annual_amount is None:
            print("Error: Annual amount must be provided for annuity calculation.")
            return
        print(f"Annual Constant Cash Flow Amount: {annual_amount:.2f}")

        # Present Value of an Annuity: sum(Annual_Amount / (1 + r)^n for n=1 to period)
        # This aligns with bringing all future cash flows to present value.
        annuity_pv_factor = sum(1 / ((1 + r)**n) for n in range(1, period + 1))
        total_present_value = annual_amount * annuity_pv_factor
        print(f"Calculated Annuity Present Value Factor: {annuity_pv_factor:.4f}")

    else:
        if amounts_per_year is None or len(amounts_per_year) != period:
            print("Error: Amounts for each year must be provided and match the period.")
            return
        print("Amounts for each year (variable cash flows):")

        # Calculate present value for each individual amount using discount factor 1 / (1 + r)^n
        for i in range(period):
            year = i + 1
            amount = amounts_per_year[i]
            # Standard present value discount factor
            discount_factor = 1 / ((1 + r)**year)
            present_value_year = amount * discount_factor
            total_present_value += present_value_year
            print(f"  Year {year}: Cash Flow={amount:.2f}, Discount Factor={discount_factor:.4f}, PV={present_value_year:.2f}")

    print("\n--- Results ---")
    print(f"Total Present Value of Future Cash Flows: {total_present_value:.2f}")

    net_present_value = total_present_value - investment_amount
    print(f"Net Present Value (NPV): {net_present_value:.2f}")

    if net_present_value > 0:
        print("\nRecommendation: This investment is **RECOMMENDABLE** as the Net Present Value is positive.")
    else:
        print("\nRecommendation: This investment is **NOT RECOMMENDABLE** as the Net Present Value is zero or negative.")

# --- Widgets for initial inputs ---
rate_widget = widgets.FloatText(
    value=5.0,
    description='Rate (%):',
    min=0.01,
    step=0.1,
    disabled=False
)

investment_amount_widget = widgets.FloatText(
    value=1000.0,
    description='Initial Investment Amount:',
    min=0.0,
    step=10.0,
    disabled=False
)

period_widget = widgets.IntText(
    value=5,
    description='Period (Years):',
    min=1,
    step=1,
    disabled=False
)

is_annuity_checkbox = widgets.Checkbox(
    value=True,
    description='Is the cash flow amount the same for all years (Annuity)?',
    disabled=False,
    indent=False
)

# Output widget to hold dynamic amount inputs and calculation results
output_area = widgets.Output()

# Function to update dynamic amount inputs based on checkbox and period
def update_amount_inputs(change):
    with output_area:
        clear_output()
        if is_annuity_checkbox.value:
            annual_amount_widget = widgets.FloatText(
                value=250.0,
                description='Annual Cash Flow Amount:',
                min=0.0,
                step=10.0,
                disabled=False
            )
            calculate_button = widgets.Button(description="Calculate Annuity PV")
            display(annual_amount_widget, calculate_button)

            def on_annuity_button_click(b):
                with output_area:
                    clear_output()
                    calculate_investment_value(
                        rate_widget.value,
                        investment_amount_widget.value,
                        period_widget.value,
                        True,
                        annual_amount=annual_amount_widget.value
                    )
            calculate_button.on_click(on_annuity_button_click)

        else:
            amounts_widgets = []
            for i in range(period_widget.value):
                amount_widget = widgets.FloatText(
                    value=0.0, # Default to 0, user will fill
                    description=f'Cash Flow Year {i+1}:',
                    min=0.0,
                    step=10.0,
                    disabled=False
                )
                amounts_widgets.append(amount_widget)
            calculate_button = widgets.Button(description="Calculate Discounted Cash Flow (DCF)")
            display(widgets.VBox(amounts_widgets), calculate_button)

            def on_dcf_button_click(b):
                with output_area:
                    clear_output()
                    amounts_list = [w.value for w in amounts_widgets]
                    calculate_investment_value(
                        rate_widget.value,
                        investment_amount_widget.value,
                        period_widget.value,
                        False,
                        amounts_per_year=amounts_list
                    )
            calculate_button.on_click(on_dcf_button_click)

# Observe changes in checkbox and period to dynamically update amount inputs
is_annuity_checkbox.observe(update_amount_inputs, names='value')
period_widget.observe(update_amount_inputs, names='value')

# Display initial widgets
display(rate_widget, investment_amount_widget, is_annuity_checkbox, period_widget, output_area)

# Call update_amount_inputs initially to set up the first state (either annuity or DCF inputs)
update_amount_inputs({'new': is_annuity_checkbox.value})
```
